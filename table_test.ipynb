```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame()
df = pd.read_table("estat_tran_r_mago_nm.tsv")
```

```python
df.head()
```

```python
# Rename first column to something easier
df = df.rename(columns={
    df.columns[0]: "meta"
})

# Split into separate columns
df[["freq", "tra_meas", "unit", "geo"]] = df["meta"].str.split(",", expand=True)

# Drop the original combined column
df = df.drop(columns=["meta"])

```

```python
df_long = df.melt(
    id_vars=["freq", "tra_meas", "unit", "geo"],
    var_name="year",
    value_name="value"
)
```
