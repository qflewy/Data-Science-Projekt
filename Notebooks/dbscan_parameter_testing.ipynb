```python
# Load cluster labels
clusters = (
    pl.read_csv(
        OUTPUT_ROOT + f"\\stations_clusters_eps{eps1}_min_samples{min_samples1}.csv"
    )
    .rename({"uuid": "station_uuid"})
    .select(["station_uuid", "cluster"])
    .lazy()
)
```

```python
# Load all daily station price data

daily_prices = pl.scan_parquet(
    str(Path(OUTPUT_PRICES) / "station_daily_mean_and_median_by_month" / "**/*.parquet")
)

print(daily_prices)
```

```python
df = daily_prices.join(
    clusters,
    on="station_uuid",
    how="left"
)
```

```python
df = df.with_columns(
    pl.when(pl.col("cluster") == -1)
    .then(pl.lit("noise"))
    .otherwise(pl.lit("cluster"))
    .alias("group")
)
```

```python
start_year = 2014
end_year = 2024

df_years = df.filter(
    (pl.col("day").dt.year() >= start_year) &
    (pl.col("day").dt.year() <= end_year)
)
```

```python
FUEL_TYPE = "e5"
daily_groups = (
    df_years
    .group_by(["day", "group"])
    .agg([
        pl.col(FUEL_TYPE + "_mean").mean().alias(FUEL_TYPE + "_mean_price"),
        pl.col(FUEL_TYPE + "_median").mean().alias(FUEL_TYPE + "_median_price")
    ])
    .sort("day")
    .collect()
)
```

```python
import plotly.graph_objects as go

df_plot = daily_groups.to_pandas()

fig = go.Figure()

# Mean Linien
for g in df_plot['group'].unique():
    subset = df_plot[df_plot['group'] == g]
    fig.add_trace(go.Scatter(
        x=subset['day'],
        y=subset[FUEL_TYPE + "_mean_price"],
        mode='lines',
        name=f"{g} mean",
        line=dict(dash='solid')
    ))

# Median Linien
for g in df_plot['group'].unique():
    subset = df_plot[df_plot['group'] == g]
    fig.add_trace(go.Scatter(
        x=subset['day'],
        y=subset[FUEL_TYPE + "_median_price"],
        mode='lines',
        name=f"{g} median",
        line=dict(dash='dot')
    ))

fig.update_layout(
    title=f"{FUEL_TYPE.upper()} Prices (Mean & Median): Cluster vs Noise ({start_year}-{end_year})",
    xaxis_title="Date",
    yaxis_title=f"{FUEL_TYPE.upper()} Price (€)",
    template="plotly_white",
    hovermode="x unified"
)

fig.show()
```
