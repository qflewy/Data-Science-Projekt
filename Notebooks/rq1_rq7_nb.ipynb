# Are there systematic (daily) cycles for fuel prices in Germany?
# AND when is the best time to fuel up
### - are there systematic daily cycles for price increases or decreases

```python
import polars as pl
import plotly.express as px
import plotly.graph_objects as go

df_changes = pl.read_parquet("D:/data/derived/hourly_price_changes_all.parquet")

plot1 = (
    df_changes
    .group_by("hour")
    .agg(pl.sum("number_of_price_changes").alias("total_changes"))
    .sort("hour")
    .to_pandas()
)

fig1 = px.line(
    plot1,
    x="hour",
    y="total_changes",
    markers=True,
    title="price changes per hour",
    labels={
        "hour": "hour of day",
        "total_changes": "number price changes per hour"
    }
)
fig1.update_layout(
    xaxis=dict(tickmode="linear", dtick=1),
    template="plotly_dark"
)
fig1.show()
```

```python

plot2 = (
    df_changes
    .group_by("hour")
    .agg([
        pl.sum("increase").alias("increase"),
        pl.sum("decrease").alias("decrease")
    ])
    .sort("hour")
    .to_pandas()
)
fig2 = go.Figure()

fig2.add_trace(
    go.Scatter(
        x=plot2["hour"],
        y=plot2["increase"],
        mode="lines+markers",
        name="Price increases"
    )
)

fig2.add_trace(
    go.Scatter(
        x=plot2["hour"],
        y=plot2["decrease"],
        mode="lines+markers",
        name="Price decreases"
    )
)

fig2.update_layout(
    title="Price changes per hour (increases vs decreases)",
    xaxis_title="Hour of Day",
    yaxis_title="Number of Price Changes",
    xaxis=dict(tickmode="linear", dtick=1),
    template="plotly_dark"
)

fig2.show()
```

## Intraday Fuel Price Cycle (Normalized)

To eliminate differences in the general price level between individual gas stations, the price index is normalized for each station:

$$P^{norm}_t = P_t - \bar{P}_{station}$$

where $\bar{P}_{station}$ is the average price index of that specific station over the observation period.

Finally, the normalized price is averaged across all stations based on the **hour of the day** ($h$) and the **day of the week** ($d$):

$$\bar{P}_{h,d} = \frac{1}{N_{h,d}} \sum P^{norm}_{h,d}$$

The y-axis thus represents the **average deviation from the station-specific price level**.  
Positive values indicate above-average prices, while negative values indicate below-average prices for that particular time of day.

```python
import polars as pl
import plotly.express as px

fuel_type = "diesel"
year = 2025
month = 10

df = pl.read_parquet(
    f"D:/data/derived/price_change_events/price_change_events_{year}_{month:02d}.parquet"
)
df = df.filter(pl.col("fuel_type") == fuel_type)
df = df.sort(["station_id", "fuel_type", "timestamp"])
df = df.with_columns(
    pl.col("price_change")
    .cum_sum()
    .over(["station_id", "fuel_type"])
    .alias("price_index")
)

df = df.with_columns(
    (
        pl.col("price_index")
        - pl.col("price_index").mean().over(["station_id", "fuel_type"])
    ).alias("price_norm")
)
hour_weekday_price = (
    df.group_by(["weekday", "hour"])
    .agg(pl.mean("price_norm").alias("mean_price"))
    .sort(["weekday", "hour"])
)
hour_weekday_price = hour_weekday_price.with_columns(
    pl.when(pl.col("weekday") == 1).then(pl.lit("Mon"))
    .when(pl.col("weekday") == 2).then(pl.lit("Tue"))
    .when(pl.col("weekday") == 3).then(pl.lit("Wed"))
    .when(pl.col("weekday") == 4).then(pl.lit("Thu"))
    .when(pl.col("weekday") == 5).then(pl.lit("Fri"))
    .when(pl.col("weekday") == 6).then(pl.lit("Sat"))
    .when(pl.col("weekday") == 7).then(pl.lit("Sun"))
    .otherwise(pl.lit("Unknown"))
    .alias("weekday_label")
)
plot_df = hour_weekday_price.to_pandas()

fig = px.line(
    plot_df,
    x="hour",
    y="mean_price",
    color="weekday_label",
    markers=True,
    title=f"Intraday Fuel Price Cycle (Normalized) - {fuel_type.upper()}"
)
fig.update_layout(
    xaxis_title="Hour of Day",
    yaxis_title="Normalized Price",
    legend_title="Weekday"
)
fig.show()
```

```python
import plotly.express as px

plot_df = hour_weekday_price.to_pandas()

order = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]

pivot_df = plot_df.pivot(index="weekday_label", columns="hour", values="mean_price")
pivot_df = pivot_df.reindex(order)

fig = px.imshow(
    pivot_df,
    aspect="auto",
    color_continuous_scale="RdBu_r",
    origin="lower",
    title=f"Intraday Fuel Price Cycle (Normalized) - {fuel_type.upper()}"
)

fig.update_layout(
    xaxis_title="Hour of Day",
    yaxis_title="Weekday",
    coloraxis_colorbar_title="Normalized Price"
)

fig.show()
```

```python
import polars as pl
import plotly.graph_objects as go
import pandas as pd

# -----------------------------
# 1. Load data
# -----------------------------
df = pl.read_parquet(
    f"D:/data/derived/price_change_events/price_change_events_{year}_{month:02d}.parquet"
)

# -----------------------------
# 2. Filter fuel type
# -----------------------------
df = df.filter(pl.col("fuel_type") == fuel_type)

# -----------------------------
# 3. Sort for cumulative reconstruction
# -----------------------------
df = df.sort(["station_id", "fuel_type", "timestamp"])

# -----------------------------
# 4. Reconstruct relative price index
# -----------------------------
df = df.with_columns(
    pl.col("price_change")
    .cum_sum()
    .over(["station_id", "fuel_type"])
    .alias("price_index")
)

# -----------------------------
# 5. Normalize within station
# -----------------------------
df = df.with_columns(
    (
        pl.col("price_index")
        - pl.col("price_index").mean().over(["station_id", "fuel_type"])
    ).alias("price_norm")
)

# -----------------------------
# 6. Aggregate by weekday and hour
# -----------------------------
hour_weekday_price = (
    df.group_by(["weekday", "hour"])
    .agg(pl.mean("price_norm").alias("mean_price"))
    .sort(["weekday", "hour"])
)

# -----------------------------
# 7. Convert to pandas pivot table
# -----------------------------
plot_df = hour_weekday_price.to_pandas()

weekday_map = {
    1: "Mon",
    2: "Tue",
    3: "Wed",
    4: "Thu",
    5: "Fri",
    6: "Sat",
    7: "Sun",
}

plot_df["weekday_label"] = plot_df["weekday"].map(weekday_map)

weekday_order = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

pivot_df = (
    plot_df.pivot(index="weekday_label", columns="hour", values="mean_price")
    .reindex(weekday_order)
)

# -----------------------------
# 8. Axes for surface plot
# -----------------------------
x = pivot_df.columns.tolist()                  # hours 0-23
y = list(range(len(pivot_df.index)))           # 0-6 for weekdays
z = pivot_df.values

# -----------------------------
# 9. Create 3D surface plot
# -----------------------------
fig = go.Figure(
    data=[
        go.Surface(
            x=x,
            y=y,
            z=z,
            colorscale="RdBu_r",
            colorbar=dict(title="Normalized Price")
        )
    ]
)

fig.update_layout(
    title=f"3D Intraday Fuel Price Cycle (Normalized) - {fuel_type.upper()}",
    scene=dict(
        xaxis=dict(
            title="Hour of Day",
            tickmode="array",
            tickvals=list(range(24))
        ),
        yaxis=dict(
            title="Weekday",
            tickmode="array",
            tickvals=list(range(7)),
            ticktext=weekday_order
        ),
        zaxis=dict(
            title="Normalized Price"
        )
    ),
    margin=dict(l=0, r=0, b=0, t=50)
)

fig.show()
```

```python
import polars as pl
import plotly.graph_objects as go
import pandas as pd

fuel_type = "diesel"
year = 2016
month = 12

# -----------------------------
# 1. Load data
# -----------------------------
df = pl.read_parquet(
    f"D:/data/derived/price_change_events/price_change_events_{year}_{month:02d}.parquet"
)

# -----------------------------
# 2. Filter fuel type
# -----------------------------
df = df.filter(pl.col("fuel_type") == fuel_type)

# -----------------------------
# 3. Sort and reconstruct price index
# -----------------------------
df = df.sort(["station_id", "fuel_type", "timestamp"])

df = df.with_columns(
    pl.col("price_change")
    .cum_sum()
    .over(["station_id", "fuel_type"])
    .alias("price_index")
)

# -----------------------------
# 4. Normalize within station
# -----------------------------
df = df.with_columns(
    (
        pl.col("price_index")
        - pl.col("price_index").mean().over(["station_id", "fuel_type"])
    ).alias("price_norm")
)

# -----------------------------
# 5. Aggregate by weekday and hour
# -----------------------------
hour_weekday_price = (
    df.group_by(["weekday", "hour"])
    .agg(pl.mean("price_norm").alias("mean_price"))
    .sort(["weekday", "hour"])
)

plot_df = hour_weekday_price.to_pandas()

weekday_map = {
    1: "Mon",
    2: "Tue",
    3: "Wed",
    4: "Thu",
    5: "Fri",
    6: "Sat",
    7: "Sun",
}
plot_df["weekday_label"] = plot_df["weekday"].map(weekday_map)

weekday_order = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
weekday_to_y = {day: i for i, day in enumerate(weekday_order)}
plot_df["ypos"] = plot_df["weekday_label"].map(weekday_to_y)

# -----------------------------
# 6. Build spike coordinates
# -----------------------------
line_x = []
line_y = []
line_z = []

for _, row in plot_df.iterrows():
    x = row["hour"]
    y = row["ypos"]
    z = row["mean_price"]

    line_x.extend([x, x, None])
    line_y.extend([y, y, None])
    line_z.extend([0, z, None])

# -----------------------------
# 7. Create figure
# -----------------------------
fig = go.Figure()

# vertical spikes
fig.add_trace(
    go.Scatter3d(
        x=line_x,
        y=line_y,
        z=line_z,
        mode="lines",
        line=dict(width=4, color="gray"),
        showlegend=False,
        hoverinfo="none"
    )
)

# top markers
fig.add_trace(
    go.Scatter3d(
        x=plot_df["hour"],
        y=plot_df["ypos"],
        z=plot_df["mean_price"],
        mode="markers",
        marker=dict(
            size=6,
            color=plot_df["mean_price"],
            colorscale="RdBu_r",
            colorbar=dict(title="Normalized Price"),
            showscale=True
        ),
        text=[
            f"Weekday: {wd}<br>Hour: {hr}<br>Normalized Price: {mp:.4f}"
            for wd, hr, mp in zip(
                plot_df["weekday_label"],
                plot_df["hour"],
                plot_df["mean_price"]
            )
        ],
        hovertemplate="%{text}<extra></extra>",
        name="Mean Price"
    )
)

fig.update_layout(
    title=f"3D Spike Plot - Intraday Fuel Price Cycle (Normalized) - {fuel_type.upper()}",
    scene=dict(
        xaxis=dict(
            title="Hour of Day",
            tickmode="array",
            tickvals=list(range(24))
        ),
        yaxis=dict(
            title="Weekday",
            tickmode="array",
            tickvals=list(range(7)),
            ticktext=weekday_order
        ),
        zaxis=dict(
            title="Normalized Price"
        )
    ),
    margin=dict(l=0, r=0, b=0, t=50)
)

fig.show()
```

```python
import polars as pl
import plotly.graph_objects as go

fuel_type = "diesel"

df = pl.read_parquet("D:/data/derived/daily_price_range_combined/daily_price_range_2024.parquet")

heat = (
    df.filter(pl.col("fuel_type") == fuel_type)
    .with_columns([
        pl.col("date").dt.month().alias("month"),
        pl.col("date").dt.weekday().alias("weekday")
    ])
    .group_by(["weekday", "month"])
    .agg(
        pl.mean("daily_range").alias("avg_daily_range")
    )
    .sort(["weekday", "month"])
)

weekday_order = [1, 2, 3, 4, 5, 6, 7]
month_order = list(range(1, 13))

z = []
for wd in weekday_order:
    row = []
    for m in month_order:
        val = (
            heat.filter((pl.col("weekday") == wd) & (pl.col("month") == m))
            .select("avg_daily_range")
        )
        row.append(val.item() if val.height > 0 else None)
    z.append(row)

weekday_labels = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

fig = go.Figure(
    data=go.Heatmap(
        x=month_order,
        y=weekday_labels,
        z=z,
        colorscale="Viridis"
    )
)

fig.update_layout(
    title=f"Average Daily Price Range: Month × Weekday - {fuel_type.upper()}",
    xaxis_title="Month",
    yaxis_title="Weekday"
)

fig.show()
```
