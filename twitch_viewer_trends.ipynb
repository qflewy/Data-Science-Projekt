```python
import requests
import pandas as pd

url = "https://sullygnome.com/api/charts/linecharts/getconfig/GameViewers/90/0/3/Counter-Strike/%20/%20/0/0/%20/0/"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "X-Requested-With": "XMLHttpRequest",
    "Referer": "https://sullygnome.com/game/counter-strike_global_offensive/90/summary"
}

response = requests.get(url, headers=headers)

data = response.json()

datasets = data["data"]["datasets"]
labels = pd.to_datetime(data["data"]["labels"])

frames = []

for dataset in data["data"]["datasets"]:
    df_temp = pd.DataFrame({
        "date": labels,
        "value": dataset["data"],
        "metric": dataset["label"]
    })
    frames.append(df_temp)

df = pd.concat(frames)

df.to_csv("counter_strike_all_metrics.csv", index=False)
```

```python
import requests

def get_cs2_players():
    # Die App-ID 730 steht für CS2
    app_id = "730"
    url = f"https://api.steampowered.com/ISteamUserStats/GetNumberOfCurrentPlayers/v1/?appid={app_id}"
    
    try:
        # Anfrage an Steam senden
        response = requests.get(url)
        response.raise_for_status() # Prüft auf HTTP-Fehler
        
        data = response.json()
        
        if data['response']['result'] == 1:
            count = data['response']['player_count']
            return count
        else:
            return "Fehler: Steam hat keine Daten geliefert."
            
    except Exception as e:
        return f"Verbindungsfehler: {e}"

if __name__ == "__main__":
    print("--- CS2 Live Tracker ---")
    spieler = get_cs2_players()
    print(f"Aktuelle Spieler online: {spieler}")
```
